```python
import os
from openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
api_key = os.getenv("API_KEY")
base_url = os.getenv("BASE_URL")
```

```python
client = OpenAI(
    api_key = api_key,
    base_url = base_url
)
```

```python
response = client.responses.create(
    model = 'qwen3.7-plus',
    input = '你好，介绍下你自己'
)
```

```python
print(response.output_text)
```

```python
stream = client.responses.create(
    model = "qwen3.7-plus",
    input = "用100字简单介绍下人工智能技术",
    stream = True
)
# with open("log.json", 'w') as f:
    
#     for event in stream:
#         f.write(event.model_dump_json(indent=2)+'\n')
    # print(event.model_dump_json())
    
    # if event.type == "response.reasoning_summary_text.delta":
    #     print("Start thinking:\n")
    #     print(event.delta)

for event in stream:
    if event.type == "response.reasoning_summary_text.delta":
        # print("Start thinking:\n")
        print(event.delta, end = '', flush=True)
    elif event.type == "response.reasoning_summary_text.done":
        print("Finishing thinking.\n")
    elif event.type == "response.output_text.delta":
        # print("Starting Answering:\n")
        print(event.delta, end = '', flush=True)
    elif event.type == "response.completed":
        print("Finishing Streaming Output.\n")
        print("Usage:", event.response.usage.total_tokens)
```
